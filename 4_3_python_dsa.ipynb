**Data Structures**

- Tuple
- Set

- Mutable

```python
# creating a list
fruits = ["apple", "banana", "mango"]
print(fruits)
```

```python
# Accessing items using index
print(fruits[0])  # prints the first element of the list
```

```python
# Change an element
fruits[2] = "orange"
print(fruits)  # prints the updated list
```

```python
a = [1, 2, 3.5, False, "word"]
print(a)  # prints the list with mixed data types
```

```python
a.append("apple")
print(a)
```

```python
a.remove("word")
print(a)
```

```python
for i in range(0,10):
    print(i)
```

```python
for num in range(0,10):
    print(num)
```

```python
#loops through a list
fruits = ["apple", "banana", "mango"]
for fruits in fruits:
    print(fruits)
```

```python
colours = ("red", "green", "blue")
print(colours[0])  # prints the first element of the list
print(type(colours))  # prints the type of the variable, which is a list
```

```python
#converting to a tuple
a = [1, 2, 3.5, False, "word"]
b = tuple(a)
print(type(b))
print(type(a))

print(b)
print(a)
```

```python
import numpy as np
a = np.array([1, 2, 3.5])
b = np.array([2,4,5])

print("Addition",a+b)
print("Subtraction",a-b)
print("Multiplication",a*b)
print("Division",a/b)
print("Cross Product",np.cross(a,b))
```

**Reshaping and Flattening**

```python
arr = np.arange(1,10) 
reshaped = arr.reshape(3,3)
flattened = reshaped.flatten()
print(arr)
print("-----------")
print(reshaped)
print(flattened)
```
