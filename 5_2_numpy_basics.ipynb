```python
import time
import numpy as np

n = 1000000

# ---------------- Python List ----------------
t1 = time.time()

py_list = [i**2 for i in range(n)]

t2 = time.time()

print("Python List first 10:", py_list[:10])
print("Python List Time:", t2 - t1)


# ---------------- NumPy Array ----------------
t1 = time.time()

np_array = np.arange(n)**2

t2 = time.time()

print("NumPy Array first 10:", np_array[:10])
print("NumPy Array Time:", t2 - t1)
```
