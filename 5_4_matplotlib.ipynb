```python
import numpy as np
import pandas as py
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("Simple Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

Customized Line plot

```python
plt.figure(figsize=(8, 6))
plt.plot(x, y, color='red', linestyle='-', linewidth=2, label='Sine Wave', marker='o', markersize=4, markerfacecolor='blue', markeredgecolor='black')
plt.title("Customized Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.grid(True)
plt.show()
```

```python
students = ['Alice', 'Bob', 'Charlie', 'David']
scores = [85, 92, 78, 90]

plt.bar(students, scores, color='skyblue')
plt.title("Student Scores")
plt.xlabel("Students")
plt.ylabel("Scores")
```

```python
students = ['Alice', 'Bob', 'Charlie', 'David']
scores = [85, 92, 78, 90]

plt.barh(students, scores, color='skyblue')
plt.title("Student Scores")
plt.xlabel("Students")
plt.ylabel("Scores")
```

Scatter plot

```python
#np.random.seed(42)
x = np.random.randn(1000)
y = np.random.randn(1000)

print((x[:5]))
print((y[:5]))
plt.scatter(x, y, color='blue', alpha=0.7)
plt.title("Scatter Plot")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

Histogram

```python
data = np.random.randn(1000)
plt.hist(data, bins=30, color="green", edgecolor='black', alpha=0.7)
plt.title("Data Distribution Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

Pie Chart


```python
labels = ['A', 'B', 'C', 'D']
sizes = [15, 30, 45, 10]
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90, colors=['gold', 'lightcoral', 'lightskyblue', 'lightgreen'])
plt.title("Pie Chart Example")
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.  
```

```python
z = np.random.randn(1000)
plt.boxplot(z, vert=True, patch_artist=True, boxprops=dict(facecolor='lightblue', color='blue'), medianprops=dict(color='red'))
plt.title("Box Plot Example")
plt.xlabel("Value")
```
